```python
import databricks.automl as automl
from utils.config import TARGET_COLUMN, AUTOML_CONFIG
from utils.validation import validate_dataframe
from utils.data_loader import load_csv


```

```python
df = load_csv("/FileStore/data/train.csv")  # Aquí se carga la data

validate_dataframe(df, TARGET_COLUMN)
```

```python
summary = automl.classify(
    dataset=df,
    target_col=TARGET_COLUMN,
    timeout_minutes=AUTOML_CONFIG["timeout_minutes"],
    primary_metric=AUTOML_CONFIG["primary_metric"]
)
```

```python
print(summary)
```
